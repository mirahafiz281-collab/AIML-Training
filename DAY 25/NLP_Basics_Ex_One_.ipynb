```python
# Step 1: Import required library
import nltk #natural language toolkit
import spacy # module NLP library
from nltk.corpus import stopwords #read file
from nltk.tokenize import word_tokenize, sent_tokenize  #pisahkan setiap ayat kpd unit kecil yang bermakna
from nltk.stem import PorterStemmer #buang imbuhan awal dan imbuhan akhiran (tinggal kata dasar je)
from nltk.stem import WordNetLemmatizer
import re
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
# Step 2: Text
our_text="Artificial Intelligence and Machine Learning are going to change the future of technology!"
```

```python
#Step 3: split text into individual words/tokens
tokens=word_tokenize(our_text)
tokens
```

```python
#Step 4: Clean token (Buang Uppercase to lower case and buang punctuation !)
clean_tokens=[re.sub(r'[^a-zA-Z]','',token.lower()) for token in tokens if token.isalpha()]
clean_tokens
```

```python
stop_words=set(stopwords.words('english'))
filtered_tokens=[token for token in clean_tokens if token not in stop_words]
filtered_tokens
```

```python
#Stemming
stemmer=PorterStemmer()
stemmed_tokens=[stemmer.stem(token) for token in filtered_tokens]
#Reduce the words to their root form (not always a word from dictionary)
stemmed_tokens
```

```python
#Lemmetization
lemmatizer=WordNetLemmatizer()
lemmatized_tokens=[lemmatizer.lemmatize(token) for token in filtered_tokens]
#convert words to their dictionary form (lemma) using POS: part of speech
lemmatized_tokens
```

```python
from nltk import corpus
corpus=[
    "Artificial Intelligence is transforming Industries",
    "Machine Learning is a branch of artificial intelligence",
    "Deep Learning is a subfield of machine learning",
    "Natural Language Processing is a subfield of artificial intelligence",
]
```

```python
cv=CountVectorizer()
bow_matrix=cv.fit_transform(corpus)
print("****Features Names (BOW)****")
print(cv.get_feature_names_out())
print("****BOW Matrix Array ****")
print(bow_matrix.toarray())
```
