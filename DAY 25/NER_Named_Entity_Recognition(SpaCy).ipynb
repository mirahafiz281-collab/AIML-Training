```python
import spacy
nlp=spacy.load("en_core_web_sm")
print('SpaCy Loaded Sucessfully')
```

```python
text='Bill Gates founded Microsoft in year 1994 in California -USA'
```

```python
doc=nlp(text)
for ent in doc.ents:
  print(ent.text,ent.label_)
```

```python
doc=nlp(text)
for ent in doc.ents:
  print(f"\n Text: {ent.text}, Label: {ent.label_},Explaination: {spacy.explain(ent.label)} \n")
```

```python
corpus=[
    "TajMahal is located in Agra India",
    "Google was founded by Larry Page",
    "Twin Towers is located in Kuala Lumpur",
    "Discovery of India was written by Pt. Nehru"
]
```

```python
for text in corpus:
  doc=nlp(text)
  print(f'\n Text: {text}')
  for ent in doc.ents:
    print(f'{ent.text} -->{ent.label_}')
```
